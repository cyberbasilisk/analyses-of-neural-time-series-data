```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.io import loadmat
from scipy.spatial import KDTree
%matplotlib inline
```

```python
# Load Data

data = loadmat('/v1_laminar.mat')
data.keys()
```

```python
# Extract Data

eeg_data = data['csd']
sampling_rate = data['srate'][0][0]
time_vector = data['timevec'][0]
timepoints = len(time_vector)

eeg_data.shape, sampling_rate, timepoints, time_vector.shape
```

```python
# Visualizing Data From A Single Channel

channel = 7

plt.close('all')
plt.subplots(1, 1, figsize=(10, 4))
plt.plot(time_vector, eeg_data[channel])
plt.xlabel('Time (s)')
plt.ylabel('Amplitude (uV)')
plt.title(f'EEG Data from channel {channel+1}')
plt.show()
```

```python
# Event Related Potential

erp = np.mean(eeg_data, axis=2)

erp.shape
```

```python
# Visualizing ERP From Same Single Channel

plt.close('all')
plt.subplots(1, 1, figsize=(10, 4))
plt.plot(time_vector, erp[channel])
plt.xlabel('Time (s)')
plt.ylabel('Amplitude (uV)')
plt.title(f' ERP from channel {channel+1}')
plt.show()
```
